```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class ProductState(TypedDict):
    price: float
    discount_percent: float
    tax: float 
    discounted_price: float
    tax_amount: float
    final_price: float
```

```python
def apply_discount(state: ProductState) -> ProductState:
    price = state["price"]
    discount = state["discount_percent"]

    discount_amount = price*(discount/100)
    discounted_price = price - discount_amount
    return {"discounted_price": discounted_price}
```

```python
def calculate_tax(state: ProductState) -> ProductState:
    discounted_price = state["discounted_price"]
    tax = state["tax"]
    tax_amount = discounted_price *(tax/100)
    return {"tax_amount":tax_amount}
```

```python
def final_price(state: ProductState)-> ProductState:
    discounted_price = state["discounted_price"]
    tax_amount = state["tax_amount"]
    final_price = discounted_price + tax_amount
    return {"final_price": final_price}
```

```python
workflow = StateGraph(ProductState)
```

```python
workflow.add_node("apply_discount", apply_discount)
workflow.add_node("calculate_tax",calculate_tax)
workflow.add_node("final_price", final_price)
```

```python
workflow.add_edge(START,"apply_discount")
workflow.add_edge("apply_discount","calculate_tax")
workflow.add_edge("calculate_tax","final_price")
workflow.add_edge("final_price",END)
```

```python
app = workflow.compile()
```

```python
initial_state = {
    "price": 1000,
    "discount_percent": 10,
    "tax": 18
}
```

```python
result = app.invoke(initial_state)
print(result)
```
