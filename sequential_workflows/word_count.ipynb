```python
from langgraph.graph import StateGraph, START, END
from pydantic import BaseModel
```

```python
### Define state using Pydantic BaseModel
class wordState(BaseModel):
    text: str
    word_count: int
    sentence: str
```

```python
def count_words(state: wordState) -> wordState:
    text = state.text
    word_count = len(text.split())
    return {"word_count": word_count}
```

```python
def sentence_length(state: wordState) -> wordState:
    count = state.word_count
    if count <5:
        sentence = "Short"
    elif count < 10:
        sentence = "Medium"
    else:
        sentence = "Long"
    return {"sentence": sentence}
```

```python
### define the graph
graph = StateGraph(wordState)

### add nodes to the graph
graph.add_node("count_words", count_words)
graph.add_node("sentence_length", sentence_length)
```

```python
## add edegs to your graph
graph.add_edge(START,"count_words")
graph.add_edge("count_words","sentence_length")
graph.add_edge("sentence_length",END)
```

```python
workflow = graph.compile()
```

```python
initial_state = {'text': "This is a test sentence for word count.",
                 "word_count": 0,
                 "sentence": ""}
final_state = workflow.invoke(initial_state)
```
